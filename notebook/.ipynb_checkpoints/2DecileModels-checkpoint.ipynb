```python
trend_path = os.path.join(data_path, "GenericTrendSignal.parquet")
df_data    = (pd
    .read_parquet(path = trend_path, engine = "pyarrow")
    .assign(signal_name = lambda x: x.Ticker + "_" + x.window.astype(str)))
```

# Decile Models

```python
df_data.head(3)
```

```python
quintile_fitter = (QuantileFitter(
    rtn_col    = "rtn", 
    group_col  = "signal_name",
    date_col   = "Date"))
```

```python
df_full_sample = quintile_fitter.fit_quantile_full_sample(df_data, "signal")
```

```python
(df_full_sample
    ["data"]
    .assign(signal_rtn = lambda x: np.where(x.signal_rtn == 0, np.nan, x.signal_rtn))
    .dropna()
    .pivot(index = "Date", columns = "signal_name", values = "signal_rtn")
    .apply(lambda x: x * (0.1 / (x.ewm(span = 100, adjust = False).std() * np.sqrt(252))))
    .mean(axis = 1).to_frame(name = "tmp")
    .apply(lambda x: np.cumprod(1 + x))
    .plot(logy = True, legend = False))
```

```python
model = quintile_fitter.fit_quantile_train_test_ratio(df_data, "signal")
```

```python
a = list(model.keys())[0]
```

```python
df_train = (pd
    .concat([model
        [name]
        ["train"]
        .assign(sample = "train")
        for name in model.keys()]))

df_test = (pd
    .concat([model
        [name]
        ["test"]
        .assign(sample = "test")
        for name in model.keys()]))

df_train_test = (pd.concat([df_train, df_test]))
```

```python
df_full_sample = quintile_fitter.fit_quantile_full_sample(df_data, "signal")["data"].assign(sample = "full_sample")
```

```python
df_expanding = quintile_fitter.fit_expanding(df_data, "signal")
```

```python
strat_name = "AGG_128"

df_tmp_expanding = (df_expanding
    .data
    .loc[lambda x: x.signal_name == strat_name]
    .set_index("Date")
    [["rtn_vol", "vol_norm_rtn"]]
    .assign(sample = "expanding"))
```

```python
df_tmp_train_test = (df_train_test
    .loc[lambda x: x.signal_name == strat_name]
    [["signal_rtn", "sample"]])
```

```python
df_full_sample = (df_full_sample
    .loc[lambda x: x.signal_name == strat_name]
    .set_index("Date")[["signal_rtn", "sample"]])
```

```python
(df_full_sample
    .drop(columns = ["sample"])
    .rename(columns = {"signal_rtn": "full_sample"})
    .merge(right = df_tmp_train_test, how = "outer", on = ["Date"])
    .drop(columns = ["sample"])
    .cumsum()
    .plot())
```
